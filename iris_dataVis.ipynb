```python
"""
Iris Dataset Analysis - Complete Solution
This script analyzes the Iris dataset with proper error handling and package installation
"""

# Step 1: Install required packages (uncomment if needed)
"""
!pip install pandas numpy matplotlib seaborn ucimlrepo
"""

# Step 2: Import libraries with error handling
import sys
import subprocess

def install_package(package):
    """Install a package using pip"""
    subprocess.check_call([sys.executable, "-m", "pip", "install", package])

# Try importing, install if missing
try:
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from ucimlrepo import fetch_ucirepo
    import warnings
    warnings.filterwarnings('ignore')
except ModuleNotFoundError as e:
    print(f"Missing module: {e.name}")
    print("Installing required packages...")
    
    # Install missing packages
    packages = ['pandas', 'numpy', 'matplotlib', 'seaborn', 'ucimlrepo']
    for pkg in packages:
        try:
            install_package(pkg)
        except:
            print(f"Failed to install {pkg}. Please install manually using:")
            print(f"pip install {pkg}")
    
    # Try importing again
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from ucimlrepo import fetch_ucirepo
    import warnings
    warnings.filterwarnings('ignore')

# Set style for better visualizations
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("="*70)
print("IRIS DATASET ANALYSIS")
print("="*70)

# Step 3: Fetch the dataset
print("\n📊 Step 1: Fetching Iris dataset from UCI Repository...")

try:
    iris = fetch_ucirepo(id=53)
    print("✅ Dataset fetched successfully!")
except Exception as e:
    print(f"❌ Error fetching dataset: {e}")
    print("Loading alternative method...")
    # Alternative: Load from scikit-learn
    from sklearn.datasets import load_iris
    iris_data = load_iris()
    iris = type('obj', (object,), {
        'data': {'features': pd.DataFrame(iris_data.data, columns=iris_data.feature_names),
                 'targets': pd.DataFrame(iris_data.target, columns=['class'])},
        'metadata': {'name': 'Iris', 'num_instances': 150, 'num_features': 4,
                    'subject_area': 'Biology', 'associated_tasks': 'Classification'},
        'variables': pd.DataFrame({
            'name': ['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 
                    'petal width (cm)', 'class'],
            'role': ['Feature', 'Feature', 'Feature', 'Feature', 'Target'],
            'type': ['Continuous', 'Continuous', 'Continuous', 'Continuous', 'Categorical'],
            'description': ['Sepal length in cm', 'Sepal width in cm', 'Petal length in cm',
                          'Petal width in cm', 'Species of iris']
        })
    })
    # Map target to class names
    iris.data.targets['class'] = iris_data.target_names[iris_data.target]

# Extract features and targets
X = iris.data.features
y = iris.data.targets

# Combine features and target
df = pd.concat([X, y], axis=1)
print(f"\n✅ Dataset loaded successfully! Shape: {df.shape}")

# Step 4: Data Exploration
print("\n" + "="*70)
print("📊 Step 2: Data Exploration")
print("="*70)

print("\n📈 First 5 rows of the dataset:")
print(df.head())

print("\n📈 Last 5 rows of the dataset:")
print(df.tail())

print("\n📊 Basic Statistics:")
print(df.describe())

# Check for missing values
print("\n📊 Missing Values Check:")
missing_values = df.isnull().sum()
print(f"Total missing values: {missing_values.sum()}")
print("✅ No missing values found!" if missing_values.sum() == 0 else "⚠️ Missing values detected!")

# Step 5: Class Distribution Analysis
print("\n" + "="*70)
print("📊 Step 3: Class Distribution Analysis")
print("="*70)

# Count samples per class
class_counts = df['class'].value_counts()
print("\n📈 Sample count per class:")
print(class_counts)

# Calculate percentages
print("\n📈 Percentage distribution:")
for class_name, count in class_counts.items():
    percentage = (count / len(df) * 100)
    print(f"  {class_name}: {count} samples ({percentage:.1f}%)")

# Step 6: Visualization
print("\n" + "="*70)
print("📊 Step 4: Creating Visualizations")
print("="*70)

# Create figure with multiple subplots
fig = plt.figure(figsize=(16, 12))

# 1. Bar Chart - Class Distribution
ax1 = plt.subplot(2, 2, 1)
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']
bars = ax1.bar(class_counts.index, class_counts.values, color=colors, 
               edgecolor='black', linewidth=1.5)

ax1.set_title('Distribution of Iris Flower Classes', fontsize=14, fontweight='bold')
ax1.set_xlabel('Class of Iris Flower', fontsize=12)
ax1.set_ylabel('Number of Samples', fontsize=12)
ax1.set_ylim(0, 60)
ax1.grid(axis='y', alpha=0.3, linestyle='--')

# Add value labels
for bar in bars:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 1.5,
             f'{int(height)}', ha='center', va='bottom', fontsize=11, fontweight='bold')

ax1.axhline(y=50, color='red', linestyle='--', linewidth=2, alpha=0.7, 
            label='Perfect Balance (50 samples)')
ax1.legend()

# 2. Pie Chart - Class Proportions
ax2 = plt.subplot(2, 2, 2)
wedges, texts, autotexts = ax2.pie(class_counts.values, 
                                   labels=class_counts.index,
                                   autopct='%1.1f%%', 
                                   colors=colors,
                                   startangle=90,
                                   explode=(0.05, 0.05, 0.05),
                                   shadow=True)

ax2.set_title('Proportion of Iris Flower Classes', fontsize=14, fontweight='bold')
plt.setp(autotexts, size=12, weight="bold", color='white')

# 3. Scatter Plot - Sepal Length vs Petal Length
ax3 = plt.subplot(2, 2, 3)
class_colors = {'Iris-setosa': '#FF6B6B', 
                'Iris-versicolor': '#4ECDC4', 
                'Iris-virginica': '#45B7D1'}

for class_name, color in class_colors.items():
    subset = df[df['class'] == class_name]
    ax3.scatter(subset['sepal length'], subset['petal length'], 
               label=class_name, color=color, s=50, alpha=0.7, edgecolors='black')

ax3.set_xlabel('Sepal Length (cm)', fontsize=11)
ax3.set_ylabel('Petal Length (cm)', fontsize=11)
ax3.set_title('Sepal Length vs Petal Length by Class', fontsize=14, fontweight='bold')
ax3.legend()
ax3.grid(True, alpha=0.3)

# 4. Box Plot - Feature Distribution
ax4 = plt.subplot(2, 2, 4)
df_melted = df.melt(id_vars=['class'], value_vars=['sepal length', 'sepal width', 
                                                    'petal length', 'petal width'],
                    var_name='Feature', value_name='Measurement (cm)')

sns.boxplot(data=df_melted, x='Feature', y='Measurement (cm)', hue='class', 
            palette=colors, ax=ax4)
ax4.set_title('Feature Distribution by Class', fontsize=14, fontweight='bold')
ax4.set_xlabel('Feature', fontsize=12)
ax4.set_ylabel('Measurement (cm)', fontsize=12)
ax4.legend(title='Class', bbox_to_anchor=(1.05, 1), loc='upper left')
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('iris_comprehensive_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Visualizations saved as 'iris_comprehensive_analysis.png'")

# Step 7: Class Balance Assessment
print("\n" + "="*70)
print("📊 Step 5: Class Balance Assessment")
print("="*70)

are_balanced = (class_counts == 50).all()
print(f"\n📈 Are all classes perfectly balanced? {are_balanced}")

if are_balanced:
    print("\n✅ YES - The Iris dataset is PERFECTLY BALANCED!")
    print("   • Each of the 3 classes contains exactly 50 samples")
    print("   • Each class represents 33.3% of the total dataset")
    print("   • No class imbalance issues - ideal for classification tasks")
else:
    print("\n❌ NO - The dataset is NOT perfectly balanced.")
    print(f"   • Class with most samples: {class_counts.idxmax()} ({class_counts.max()} samples)")
    print(f"   • Class with least samples: {class_counts.idxmin()} ({class_counts.min()} samples)")
    print(f"   • Difference: {class_counts.max() - class_counts.min()} samples")

print(f"\n📊 Balance Metrics:")
print(f"   • Total samples: {len(df)}")
print(f"   • Number of classes: {len(class_counts)}")
print(f"   • Samples per class - Min: {class_counts.min()}, Max: {class_counts.max()}")
print(f"   • Standard deviation: {class_counts.std():.2f}")
print(f"   • Balance ratio: {class_counts.iloc[0]}:{class_counts.iloc[1]}:{class_counts.iloc[2]}")

# Step 8: Correlation Analysis
print("\n" + "="*70)
print("📊 Step 6: Correlation Analysis")
print("="*70)

feature_cols = ['sepal length', 'sepal width', 'petal length', 'petal width']
correlation_matrix = df[feature_cols].corr()

print("\n📈 Correlation Matrix:")
print(correlation_matrix)

# Create correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Feature Correlation Heatmap', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('iris_correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Correlation heatmap saved as 'iris_correlation_heatmap.png'")

# Step 9: Summary Insights
print("\n" + "="*70)
print("📊 VISUALIZATION INSIGHTS SUMMARY")
print("="*70)

print("\n1. CLASS DISTRIBUTION (Bar Chart):")
print("   • All three Iris species are equally represented with 50 samples each")
print("   • Each class represents exactly 33.3% of the dataset")
print("   • The red dashed line confirms perfect balance")

print("\n2. PROPORTION ANALYSIS (Pie Chart):")
print("   • The dataset has a perfect 1:1:1 ratio")
print("   • No class dominates or is underrepresented")

print("\n3. FEATURE RELATIONSHIPS (Scatter Plot):")
print("   • Iris Setosa is clearly separable from Versicolour and Virginica")
print("   • Petal length shows strong discriminatory power")
print("   • Versicolour and Virginica show some overlap")

print("\n4. FEATURE DISTRIBUTION (Box Plot):")
print("   • Petal features provide clear separation between species")
print("   • Sepal features show more overlap")
print("   • Iris Virginica generally has larger measurements")

print("\n5. CORRELATION ANALYSIS:")
print("   • Petal length and petal width: Strong correlation (0.96)")
print("   • Sepal length and petal length: Strong correlation (0.87)")
print("   • Sepal width: Weakly correlated with other features")

print("\n6. CLASS BALANCE CONCLUSION:")
print("   ✅ The Iris dataset is PERFECTLY BALANCED")
print("   ✅ No sampling techniques needed")
print("   ✅ Ideal for teaching classification algorithms")
print("   ✅ Model metrics won't be biased by class imbalance")

# Save summary to file
with open('iris_analysis_summary.txt', 'w') as f:
    f.write("IRIS DATASET ANALYSIS SUMMARY\n")
    f.write("="*50 + "\n\n")
    f.write(f"Total samples: {len(df)}\n")
    f.write(f"Number of classes: {len(class_counts)}\n")
    f.write("Class distribution:\n")
    for class_name, count in class_counts.items():
        f.write(f"  {class_name}: {count} samples ({count/len(df)*100:.1f}%)\n")
    f.write(f"\nPerfectly balanced: {are_balanced}\n")
    f.write(f"Balance ratio: 1:1:1\n")

print("\n✅ Analysis complete! Summary saved to 'iris_analysis_summary.txt'")
print("="*70)
```
